```python
df = pd.read_csv('Bestsellers with categories.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
Mean_User_Rating = np.mean(df['User Rating'])
print(f"Mean User rating is: {Mean_User_Rating:.1f}" )
```

```python
Mean_Price= np.mean(df['Price'])
print(f"Mean Price is: {Mean_Price:.1f}" )
```

```python
Mean_R= np.mean(df['Reviews'])
print(f"Mean review is: {Mean_R:.1f}" )
```

```python
Median_User_Rating = np.median(df['User Rating'])
print(f"Median User rating is: {Median_User_Rating:.1f}" )
```

```python
Median_Price= np.median(df['Price'])
print(f"Median Price is: {Median_Price:.1f}" )
```

```python
Median_R= np.median(df['Reviews'])
print(f"Median review is: {Median_R:.1f}" )
```

```python
mode_User_R=df['User Rating'].mode()
print(f"Mode User rating is: {mode_User_R}" )
```

```python
mode_Price=df['Price'].mode()
print(f"Mode Price is: {mode_Price}" )
```

```python
mode_Review=df['Reviews'].mode()
print(f"Mode Review is: {mode_Review}" )
```
